```python
#29-OCT-26 Eliza yepes 

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

iris = sns.load_dataset("iris")


sns.set_theme(
    style="whitegrid",
    context="notebook",
    palette="deep",
    font_scale=1.0
)

colores = ["#D494D5", "#AC2C84", "#792062"]

plt.rcParams.update({
    "figure.dpi": 110,
    "axes.titleweight": "bold",
    "axes.titlepad": 12,
    "axes.spines.top": False,
    "axes.spines.right": False
})



fig, axes = plt.subplots(
    2, 2,
    figsize=(12, 8)
)


sns.histplot(
    data=iris,
    x="sepal_length",
    kde=True,
    color=colores[0],
    ax=axes[0, 0]
)

axes[0, 0].set_title(
    "Distribución de la longitud del sépalo"
)

axes[0, 0].set_xlabel("Longitud del sépalo (cm)")
axes[0, 0].set_ylabel("Frecuencia")


sns.boxplot(
    data=iris,
    x="species",
    y="petal_length",
    color=colores[1],
    ax=axes[0, 1]
)

axes[0, 1].set_title(
    "Longitud del pétalo por especie"
)

axes[0, 1].set_xlabel("Especie")
axes[0, 1].set_ylabel("Longitud del pétalo (cm)")



sns.scatterplot(
    data=iris,
    x="petal_length",
    y="petal_width",
    hue="species",
    palette=colores,
    s=65,
    ax=axes[1, 0]
)

axes[1, 0].set_title(
    "Relación entre longitud y ancho del pétalo"
)

axes[1, 0].set_xlabel("Longitud del pétalo (cm)")
axes[1, 0].set_ylabel("Ancho del pétalo (cm)")



sns.violinplot(
    data=iris,
    x="species",
    y="sepal_width",
    hue="species",
    palette=colores,
    dodge=False,
    legend=False,
    ax=axes[1, 1]
)

axes[1, 1].set_title(
    "Distribución del ancho del sépalo por especie"
)

axes[1, 1].set_xlabel("Especie")
axes[1, 1].set_ylabel("Ancho del sépalo (cm)")


fig.suptitle(
    "Análisis de las características de las flores Iris",
    fontsize=17,
    fontweight="bold"
)

sns.despine()

plt.tight_layout(rect=[0, 0, 1, 0.96])

plt.show()
```
