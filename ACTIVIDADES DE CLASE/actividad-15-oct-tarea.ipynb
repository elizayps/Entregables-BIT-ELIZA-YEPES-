```python
import numpy as np

ventas = np.array([120, 85, 150, 200, 95, 180, 75, 220, 160, 90])
# 1. Total vendido
total = np.sum(ventas)
print("TOTAL VENDIDO")
print(total)
print("------------------------------------------------")

# 2. Promedio y mediana
promedio = np.mean(ventas)
print("Promedio")
print(promedio)
mediana = np.median(ventas)
print("Mediana")
print(mediana)
print("------------------------------------------------")

# 3. Ventas inferiores a 100
ventas < 100
ventas[ventas < 100]
dias = np.where(ventas < 100)[0] + 1
print("Ventas inferiorres a 100")
print(dias)
print("-----------------------------------------------")

# 4. Porcentaje de días con ventas superiores a 150
dias_superiores = np.sum(ventas > 150)
porcentaje = (dias_superiores / len(ventas)) * 100
print("Porcentaje de dias con ventas superiores a 100")
print(porcentaje)
print("-----------------------------------------------")

# 5. Etiquetas de alerta
etiquetas = np.where(ventas < 100, "¡ALERTA!", "NORMAL")
print("ETIQUETAS")
print(etiquetas)
print("-----------------------------------------------")

# 6. Escribe tus conclusiones en una celda Markdown.
 #las ventas de los 10 dias dio un total de 1375 u/d
 #el promedio fue de 137.5  
 #las ventas superiores en esos 10 dias fueron del 40%


```
