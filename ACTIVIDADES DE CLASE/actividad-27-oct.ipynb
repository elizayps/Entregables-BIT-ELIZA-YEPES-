```python
#actividad 27-oct-26 Eliza yepes 
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.DataFrame({
    "Mes": ["Ene", "Feb", "Mar", "Abr", "May", "Jun"],
    "Ventas": [120, 150, 135, 180, 210, 195],
    "Categoria": ["A", "B", "A", "C", "B", "C"],
    "Unidades": [20, 35, 25, 40, 30, 45]
})
```

```python
plt.figure(figsize=(10, 5))

plt.plot(
    df['Mes'],
    df['Ventas'],
    color="#44073B",
    marker="o",
    markerfacecolor="#B047A0",
    markeredgecolor="#E489D7",
    markersize=10,
    linewidth=2.5,
    linestyle="-"
)

plt.title("VENTAS POR MES", fontsize=16)

plt.xlabel("Mes", fontsize=12)

plt.ylabel("Ventas", fontsize=12)

plt.xticks(rotation=45)

plt.grid(linestyle="--", alpha=0.4)

plt.show()
```

```python
# Configurar estilo visual
sns.set_theme(style="whitegrid")

# Crear paleta personalizada
mis_colores = ["#B486D7", "#7F339A", "#3C0446"]

# Tamaño de la figura
plt.figure(figsize=(10, 5))

# Crear gráfico de barras
sns.barplot(
    data=df,
    x="Categoria",
    y="Unidades",
    hue="Categoria",
    palette=mis_colores,
    legend=False
)

# Personalizar título y etiquetas
plt.title("UNIDADES POR CATEGORÍA", fontsize=16)

plt.xlabel("Categorías", fontsize=12)

plt.ylabel("Promedio de unidades", fontsize=12)

# Personalizar cuadrícula
plt.grid(axis="y", linestyle="--", alpha=0.4)

# Mostrar gráfico
plt.show()
```
